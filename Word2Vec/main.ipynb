```python
import gensim
from gensim.models import Word2Vec, KeyedVectors  
```

```python
import gensim.downloader as api
wv = api.load('word2vec-google-news-300')
vector_king = wv['king'] #this gives us the vector for the word 'king' against the features of the word2vec model. This is a 300 dimensional vector. 
```

```python
vector_king 
```

```python
vector_king.shape #As mentioned above, this is a 300 dimensional vector. We can verify that by checking the shape of the vector.
```

```python
wv.most_similar('happy') #This gives us the words that are most similar to the word 'happy' based on the word2vec model.
```

```python
wv.similarity('happy', 'glass') #This gives us the similarity between the words 'happy' and 'sad' based on the word2vec model.
```

```python
new_vector = wv['football'] - wv['foot'] + wv['basket']
wv.most_similar(new_vector)
```
